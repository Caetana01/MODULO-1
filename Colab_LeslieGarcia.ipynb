```python
import matplotlib.pyplot as plt
a = 81
b = 12
resultado = a * b
eje_x = [0, b]
eje_y = [0, resultado]
plt.step(eje_x, eje_y, color='blue', where='post', linewidth=2)
plt.title('perdida de liquido')
plt.xlabel('segundos')
plt.ylabel('resultado de la multiplicación')
print(f"Resultado de la multiplicación: {resultado}")

plt.show()
```
